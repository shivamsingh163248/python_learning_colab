```python
def checking(num):
  for i in range(2, num):
    if num % i == 0:
      return False
  return True

def Solve(num):
  # write your code here
  # creating the reverse of the number 
  rev = int(str(num)[::-1])
  
  # now checking the prime number 
  # now calling the function 
  if checking(num) and checking(rev):
    print("Yes")
  else:
    print("No")
```
