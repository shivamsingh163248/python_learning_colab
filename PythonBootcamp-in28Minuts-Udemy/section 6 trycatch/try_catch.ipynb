```python
#multiped except class in the python 
try:
    i = 0 
    j = 10/i
    values = [1 , '1']
    print(sum(values))
except TypeError: 
    print("type error")
    i = 0
except ZeroDivisionError:
    print("zero division error ")

print(j)
print(i)
```

```python
#try using the base exception 
try:
    10/0
except BaseException:
    print("zero division error")
except TypeError:
    print(TypeError)
```

```python
#try using the custom class exception 

class mylarsException(BaseException): pass

try:
    10/0
except ZeroDivisionError:
    print("zero division error")
except mylarsException:
    print("custom error")
except TypeError:
    print(TypeError)
```

```python
class Currency:
    def __init__(self , currency , amount):
        self.currency = currency
        self.amount = amount

    def __repr__(self):
        return repr((self.currency , self.amount))
    

check = Currency(55 , "shi")
print(check)
```
