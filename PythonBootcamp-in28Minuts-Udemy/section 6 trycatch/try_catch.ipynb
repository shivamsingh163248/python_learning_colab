```python
class Currency:
    def __init__(self, currency, amount):
        self.currency = currency
        self.amount = amount

    def __repr__(self):
        return f"Currency('{self.currency}', {self.amount})"

    def __add__(self, other):
        if self.currency != other.currency:
            raise Exception("Currency mismatch! Cannot add.")
        total_amount = self.amount + other.amount
        return Currency(self.currency, total_amount)

value1 = Currency("USD", 56)
value2 = Currency("INR", 45)
value3 = Currency("USD", 65)

try:
    print(value1 + value2)  # Raises Exception
except Exception as e:
    print("Error:", e)

print(value1 + value3)      # Works
```
