```python
#multiped except class in the python 
try:
    i = 0 
    j = 10/i
    values = [1 , '1']
    print(sum(values))
except TypeError: 
    print("type error")
    i = 0
except ZeroDivisionError:
    print("zero division error ")

print(j)
print(i)
```
