```python
class nit:
    def __init__(self , name , branch):
        self.name = name
        self.branch = branch


    #creating the function 
    def prints(self):
        print(f"{self.branch} {self.name}")

    #creating the another function 
    def check():
        print("this is the department")

    def par(self, val):
        print(f"this is the val {val}")


#now creating the object 
shivam = nit("cse" , "shivam singh")

shivam.prints()
shivam.par(4)
print(shivam.branch)
print(shivam.name)
        
```
