```python
class MATER_BIKE : 
    def __init__(self , speed):
        self.speed = speed
        

    #creating the method and change the objet state through the method 
    def increment(self):
        self.speed += 1

    #creating the second function to change the state of the variable 
    def decrement(self):
        self.speed -= 1 

#set the speed and print the speed 
shi = MATER_BIKE(50) 
#noe printing the values 
print(shi.speed)
shi.speed = 20
print(shi.speed)
shi.decrement()

print(shi.speed)
shi.increment()
print(shi.speed)
```

```python
import math

class Point:

    def __init__(self, x, y): 
        self.x = x
        self.y = y

    def move(self, dx, dy):
        #adding the distance in the python 
        self.x = self.x + dx
        self.y = self.y + dy
        pass

    def distance_to(self, other):
        #the pass another object 
        return math.sqrt((other.x - self.x)+(other.y - self.y))
        pass
```

```python
class RGBColor:
    def __init__(self, red, green, blue):
        self.red = red
        self.green = green 
        self.blue = blue
        pass

    def invert(self):
        self.red = 225 - self.red
        self.green = 225 -  self.green 
        self.blue = 225 - self.blue
        pass
    

color = RGBColor(255, 0, 0)
print(color.red)
color.invert()
print(color.red)   # Prints: 0
print(color.green) # Prints: 255
print(color.blue)  # Prints: 255
```
