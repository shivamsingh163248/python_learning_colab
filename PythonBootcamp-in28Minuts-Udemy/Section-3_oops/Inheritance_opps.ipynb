```python
#creating the animal class 
class Animal :
    #animal has the property of the barking 
    def bark(self):
        print("bark")

#create a obj and print 
dog= Animal()
dog.bark()

class pet(Animal):
    #creating the method of the groom 
    def groom(self):
        print("groom")


#checking and the printing 
ped_obj = pet()
ped_obj.groom()

#now creating the inheritance property would be inherit by another class
ped_obj.bark() 
```
