```python

class LandAnimal:
    def __init__(self):
        super().__init__()
        self.walking_speed = 5
 
class WaterAnimal:
    def __init__(self):
        super().__init__()
        self.swimming_speed = 10
 
class Amphibian(WaterAnimal, LandAnimal):
    def __init__(self):
        super().__init__()



amphibian = Amphibian()
print(amphibian.swimming_speed)  # Output: 10
print(amphibian.walking_speed)  # Output: 5
```

```python
class School:
    school_class = []

    def __init__(self):
        print("This is the school")

    def add_class(self, add_class):
        self.school_class.append(add_class)


class SchoolClass:
    def __init__(self):
        print("This is the school -- class")


class Values(School, SchoolClass):
    def __init__(self):
        School.__init__(self)       # Explicitly call School constructor
        SchoolClass.__init__(self)  # Explicitly call SchoolClass constructor


val = Values()
```

```python
# Creating an abstract class
from abc import ABC , abstractmethod

class Water(ABC):
    @abstractmethod
    def weight(self):
        pass


class Cloud(Water):
    def weight(self):
        print("This is the weight")


s = Cloud()
s.weight()

```

```python
#creating the class 
from abc import ABC , abstractmethod
class Restaurant():
    def Process(self):
        self.cutting()
        self.cooking()
        self.packed()


    @abstractmethod
    def cutting(self):
        pass

    @abstractmethod
    def cooking(self):
        pass

    @abstractmethod
    def packed(self):
        pass 


#creating the implementation class 
class imp(Restaurant):

    def cutting(self):
        print("no cutting")
    
    def cooking(self):
        print("cooking")
        pass
    
    def packed(self):
        print("packed")
        pass 


va = imp()
va.Process()
```

```python
#learning the polymorphism 

class Shape:
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius * self.radius
 
class Square(Shape):
    def __init__(self, side):
        self.side = side
    
    def area(self):
        return self.side * self.side
 
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    def area(self):
        return self.length * self.width
    

shapes = [Circle(5), Square(4), Rectangle(2, 5)]
 
for shape in shapes:
    print(f"The area of {shape} is {shape.area()}")
```
