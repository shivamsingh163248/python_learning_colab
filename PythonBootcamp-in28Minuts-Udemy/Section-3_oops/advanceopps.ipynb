```python
#hear the learning the oops concept of the python 
# create a class 
class fan :
    #create a constructor 
    def __init__(self , make , radius , color):
        self.make = make
        self.radius = radius
        self.color = color
        self.speed = 0 
        self.switch = False

    def __repr__(self):
        return repr((self.make , self.radius , self.color))
    
    def switch_on(self):
        self.switch = True

    def switch_off(self):
        self.switch = False
    

#now creating the object and try to print the values 
print(fan("svm" , 45.6 , "red"))
```

```python
class Book : 
    def __init__(self , id , name , author):
        self.id = id
        self.name = name
        self.author = author
        self.reviews =[]

    # hear we creating the readable function 
    def __repr__(self):
        return repr((self.id , self.name , self.author , self.reviews))
    
    def set_review(self , review):
        self.reviews.append(review)
    

#crating the another class of the reviews
class Reviews : 
    # NOW I write a counstoctur 
    def __init__(self , rating , comment , other):
        self.rating = rating
        self.comment = comment
        self.other = other


    #now creating the readable function 
    def __repr__(self):
        return repr((self.rating , self.comment , self.other))
    

#now creating the object and try to print the both object 

book_obj = Book(454 , "computer organization architecture" , "dr . shivam singh " )
print(book_obj)
#noe creating the 
reviews_obj = Reviews(6 , "nyc product" , "required more changed ")
print(reviews_obj)

#calling the function 
book_obj.set_review(reviews_obj)
#now again print 
print(book_obj)

        

    
```
