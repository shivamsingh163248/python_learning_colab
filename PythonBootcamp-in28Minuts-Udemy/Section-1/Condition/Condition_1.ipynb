```python
def is_valid_triangle(angle1, angle2, angle3):
    if angle1 <= 0 or angle2 <= 0 or angle3 <= 0:
        return False
    elif angle1 + angle2 + angle3 == 180:
        return True
    else:
        return False

# Test the function
is_valid_triangle(30, 60, 90)  # Return True
```

```python
def calculate_sum_of_divisors(number):
    sum = 0 
    if number <= 0 :
        return sum
    
    #creating the for loop 
    for i in range(1,   number+1):
        if(number%i == 0 ):
            sum += i 
            print(i , end=" ")
          
    return sum 
    
print(calculate_sum_of_divisors(6))   # Output: 12
print(calculate_sum_of_divisors(15))  # Output: 24
```
