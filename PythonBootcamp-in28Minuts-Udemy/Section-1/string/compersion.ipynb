```python
# developed the string compersion function 
def check_comp(str1 , str2):
    #check for the every condition 
    if str1.upper() == str2.upper() :
        print("True")
    if(str1.lower() == str2.lower()):
        print("True")
    #next check for the normal 
    if(str1 == str2):
        print("True")
    else :
        print('False')
```

```python
check_comp("shivam" , "Shivam")
check_comp("shivam" , "shivam")
```

```python
import string 
def count_uppercase_letters(text):
    #creating the count upper case latter 
    count = 0
    for i in range(len(text)):
       if text[i] in string.ascii_uppercase:
           count += 1
           # TODO: write code...  
           
    return count
    
    
print(count_uppercase_letters('Hello WORLD'))  # Output: 6
```
