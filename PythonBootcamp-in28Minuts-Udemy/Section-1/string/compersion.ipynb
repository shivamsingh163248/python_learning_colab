```python
# developed the string compersion function 
def check_comp(str1 , str2):
    #check for the every condition 
    if str1.upper() == str2.upper() :
        print("True")
    if(str1.lower() == str2.lower()):
        print("True")
    #next check for the normal 
    if(str1 == str2):
        print("True")
    else :
        print('False')
```

```python
check_comp("shivam" , "Shivam")
check_comp("shivam" , "shivam")
```

```python
def find_right_most_digit(text):
    # Reverse the text and convert it to a list
    newText = list(reversed(text))
    # Print the reversed text for debugging
    #print(newText)
    # Iterate through the reversed text to find the first digit
    for i in range(len(newText)):
        if newText[i].isdigit():
            return int(newText[i])
    return -1

print(find_right_most_digit('The value is 42'))  # Output: 2
print(find_right_most_digit('No digits here'))  # Output: -1
```
