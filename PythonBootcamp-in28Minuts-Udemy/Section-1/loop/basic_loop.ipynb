```python
#creating the loop 
for i in range(1, 10):
    print(f"{i} " , end="")
```

```python
for i in range(5, 10):
    print(f"{i} " , end="")
```

```python
for i in range(10 , 5):
    print(f"{i} " , end="")
```

```python
for i in range(10, 0, -1):
    print(f"{i} ", end="")
```

```python
#creating the increment of the loop 
for i in range(3 , 10 , 2):
    print(i , end=" ")
```
