```python
#creating the set 
set1 = {35,6,54,6,5,6}
print(set1)


#now updating the set
set1.add(454) #not fellow the order 

print(set1)


# To check if 7 is in the set, use 'in' keyword
print(7 in set1)
print(35 in set1)


#removing the element 
print(set1.remove(35))
print(set1)

#creating the blank set 
set2 = set()
set2.add(5)
print(set2)

num1 = [5,7,5,6,5,7]
newSet = set(num1)
print(newSet)
#this is the set adding the same element not sowing this is not fellow the

#now test the aggregate function min , max sum , len
print(sum(newSet))
print(min(newSet))
print(max(newSet))
print(len(newSet))


#creating the testing the union and minus and range intersection 
#creating the two set 
NewNumberSet = {1,2,3,4,6,7,}
NewNumberSet2 = {5,6,7,8,9,10,11,12}

print(NewNumberSet.union(NewNumberSet2))
print(NewNumberSet|NewNumberSet2)
print(NewNumberSet.intersection(NewNumberSet2))
print(NewNumberSet&NewNumberSet2)


#trying the -
print(NewNumberSet.difference(NewNumberSet2))
print(NewNumberSet2.difference(NewNumberSet))

#trying using the operator 
print(NewNumberSet - NewNumberSet2)

#now learning the range in the 
print(NewNumberSet.pop())
print(set(range(1,5)))
```

```python
#unpacking the element 
def unpacking(Element1 , Element2 , Element3):
    print(Element1)
    print(Element2)
    print(Element3)


num = [3,5,6]
num2 = {3,6,5}

#calling the function of the packing 
unpacking(*num)
unpacking(*num2)
```
