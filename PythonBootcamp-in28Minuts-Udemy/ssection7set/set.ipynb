```python
def find_intersection(num1, num2, limit):
    set1 = {i for i in range(num1, limit, num1)}
    set2 = {j for j in range(num2, limit, num2)}
    set3 = set1 & set2
    return set3

print(find_intersection(3, 3, 30))  # Output: {12, 24}
print(find_intersection(3, 5, 20))  # Output: {15}
```

```python
def find_intersection(num1, num2, limit):
    if num1 == 0 or num2 == 0:
        return 
    
    set1 = {i for  i in range(num1 , limit+1 , num1)}
    set2 = {i for  i in range(num2, limit+1 , num2)}
    return set1 & set2
    
    
    
print(find_intersection(4, 6, 30))  # Output: {12, 24}
print(find_intersection(3, 5, 20))  # Output: {15}
```
