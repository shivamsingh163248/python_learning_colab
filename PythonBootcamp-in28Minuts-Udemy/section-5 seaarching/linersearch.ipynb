```python
numbers = [12, 34, 56, 78, 23, 45, 67, 89, 21, 43, 65, 87]
#using the for the loop apply the linear search 
#creating the function if the found the number 

def find_numbers(num):
    for i in range(len(numbers)):
        
        if numbers[i] == num:
            return i
    return "not found"
        
#calling the function to check the test case 
print(find_numbers(56))
print(find_numbers(454))


    
```

```python
# Binary search using a recursive approach
numbers = [12, 21, 23, 34, 43, 45, 56, 65, 67, 78, 87, 89]

def search_num(numbers, target, left=0, right=None):
    # Set the default value of right on the first call
    if right is None:
        right = len(numbers) - 1

    # Base condition: target not found
    if left > right:
        return "Not FOUND"

    mid = (left + right) // 2

    # If the target is found at mid
    if numbers[mid] == target:
        return mid

    # If the target is in the right half
    elif numbers[mid] < target:
        return search_num(numbers, target, left=mid + 1, right=right)

    # If the target is in the left half
    else:
        return search_num(numbers, target, left=left, right=mid - 1)

# Example usage
print(search_num(numbers, 65))  # Output should be the index of 65 in the list

    
```
