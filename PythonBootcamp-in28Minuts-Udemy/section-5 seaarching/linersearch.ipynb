```python
numbers = [12, 34, 56, 78, 23, 45, 67, 89, 21, 43, 65, 87]
#using the for the loop apply the linear search 
#creating the function if the found the number 

def find_numbers(num):
    for i in range(len(numbers)):
        
        if numbers[i] == num:
            return i
    return "not found"
        
#calling the function to check the test case 
print(find_numbers(56))
print(find_numbers(454))


    
```
