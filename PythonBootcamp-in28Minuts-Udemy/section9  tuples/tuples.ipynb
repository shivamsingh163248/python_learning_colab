```python
def swap_elements(input_tuple):
    list1 = list(input_tuple)
    #print(list1)
    left = 0 
    right = len(list1) - 1
    while(left<right):
        temp = list1[left]
        list1[left] = list1[right]
        list1[right] = temp
        #print(list1[right])
        left += 1
        right -= 1

    return tuple(list1)

print(swap_elements((1, 2, 3, 4)))          # Output: (4, 2, 3, 1)
print(swap_elements((7, 14, 21, 28)))      # Output: (28, 14, 21, 7)
print(swap_elements(('apple', 'banana', 'cherry'))) # Output: ('cherry', 'banana', 'apple')
print(swap_elements((5, 10)))               # Output: (10, 5)


```

```python
def swap_elements(input_tuple):
    if len(input_tuple) < 2:
        raise ValueError("Tuple must have at least two elements")
    # Convert to list to swap elements
    temp_list = list(input_tuple)
    temp_list[0], temp_list[-1] = temp_list[-1], temp_list[0]
    return tuple(temp_list)

print(swap_elements((1, 2, 3, 4)))          # Output: (4, 2, 3, 1)
print(swap_elements((7, 14, 21, 28)))      # Output: (28, 14, 21, 7)
print(swap_elements(('apple', 'banana', 'cherry'))) # Output: ('cherry', 'banana', 'apple')
print(swap_elements((5, 10)))               # Output: (10, 5)
```

```python
def student_summary(student):
    name , age, greade = student
    return f"Student Name: {name}, Age: {age}, Grade: {greade}"


# Example 1
print(student_summary(('Alice', 20, 89.5)))  
    # Expected Output: "Student Name: Alice, Age: 20, Grade: 89.5"
     
# Example 2
print(student_summary(('Bob', 22, 75.8)))  
    # Expected Output: "Student Name: Bob, Age: 22, Grade: 75.8"
     
    # Example 3
print(student_summary(('Charlie', 19, 92.0)))  
    # Expected Output: "Student Name: Charlie, Age: 19, Grade: 92.0"
```
