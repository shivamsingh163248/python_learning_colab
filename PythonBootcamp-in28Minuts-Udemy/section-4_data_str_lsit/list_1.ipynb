```python
# Example number
number = 12.3456789

# Rounding to 2, 1, and 3 decimal points
rounded_2 = round(number, 2)
rounded_1 = round(number, 1)
rounded_3 = round(number, 3)

print("Original number:", number)
print("Rounded to 2 decimal points:", rounded_2)
print("Rounded to 1 decimal point:", rounded_1)
print("Rounded to 3 decimal points:", rounded_3)
```
