```python
#now creating row of 2d and print 
row = 4 
col = 5

list_val = []
#creating the loop 

for j in range(row):
    rows = []
    for i in range(col):
        rows.append(i+j)  # Use append to add elements to the rows list
    list_val.append(rows)  # Use append to add rows to the list_val

print(list_val)
print(list_val[2][3])
```
