```python
list1 = ["one", "two", "three", "four", "five", "six", "seven", "eight", "nine", "ten",
         "eleven", "twelve", "thirteen", "fourteen", "fifteen", "sixteen", "seventeen", "eighteen", "nineteen", "twenty"]


#creating the function of the size 

def update_list(number):
    #now updating the list with the size of the 4 
    list2 = []
    for i in list1:
        #creating the condition 
        if len(i) == 4 :
            #now append in the list 
            list2.append(i)

    return list2 


#now calling the function 
print(update_list(list1))


```

```python
#creating the new list 

comprehension  = [ i for i in list1 if len(i) == 5]
print(comprehension)
comprehension  = [len(i) for i in list1 ]
print(comprehension)

#now creating the reverse of the list 
comprehension  = [i[::-1] for i in list1 ]
print(comprehension)

#creating the list if belong e in the list then print 
comprehension  = [ i  for i in list1 if  "e" in i  ]
print(comprehension)
```

```python
#creating the list of the even number and odd number and the prime number in the 100 

comprehension  = [ i  for i in range(100) if i%2 == 0 ]
print(comprehension)

comprehension  = [ i  for i in range(100) if i%2 != 0 ]
print(comprehension)


#creating the list of the prime number 
comprehension  = [ [for i in range(i/2)]  for i in range(100) ]
print(comprehension)
```
