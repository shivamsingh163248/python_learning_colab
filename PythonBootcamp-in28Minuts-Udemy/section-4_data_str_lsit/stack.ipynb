```python
#creating the class of the stack 
class stack:
    #creating the arrays 
    #creating the constructor 
    def __init__(self):
        self.arr = []

    #now creating the method push 
    def push(self , element):
        self.arr.append(element)

    #creating the operation of the pop and return the last element and delete 
    def pop(self):
        return self.arr.pop
    

    #find the last element 
    def last_ele(self):
        return self.arr[-1]
    

    #creating the function isEmpty
    def isEmpty(self):
        #checking for the boolean 
        return not bool(self.arr) 
```
