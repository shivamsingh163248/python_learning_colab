```python
#creating the class of the stack 
class stack:
    #creating the arrays 
    #creating the constructor 
    def __init__(self):
        self.arr = []

    #now creating the method push 
    def push(self , element):
        self.arr.append(element)

    #creating the operation of the pop and return the last element and delete 
    def pop(self):
        return self.arr.pop()
    

    #find the last element 
    def last_ele(self):

        #creating the try and except 
        try:
            return self.arr[-1]
        except IndexError as e :
            print("my stack is empty")
            print(e)
        except:
            return -1 
    

    #creating the function isEmpty
    def isEmpty(self):
        #checking for the boolean 
        return not bool(self.arr) 
    


#creating the object 
new_obj = stack()
new_obj.push(4)
new_obj.push(5)

#noe pop the element 
print(new_obj.pop())
print(new_obj.last_ele())

print(new_obj.pop())

#checking the for the empty 
print(new_obj.isEmpty())
print(new_obj.last_ele())

```
