```python
#creating the tuple 
tuple_1 = (3,5,4,67,5)
tuple_2 = ("shivam" ,5,True ,6)
```

```python
#creating the blank tuple or single tuple '
tuple_3 = (4)
print(type(tuple_3))
tuple_3 = (4,)
print(type(tuple_3))
```

```python
#creating with the empty and the nested tuple 
tuple_4 = ()
print(type(tuple_4))
```

```python
#and creating one more type the empty tuple 
tuple_4 = tuple()
print(tuple_4)
```

```python
#creating the nested tuple and the access element in the tuple 
tuple_5 = ((4,5),(47,78),(4,6,8),(4,5),(6,8,2))
print(tuple_5[3])
print(tuple_5[:3])
print(tuple_5[3:])
print(tuple_5[-1])
print(tuple_5[2][2])


```

```python
#using ::-1 reversing the tuples 
print(tuple_5[::-1])
```

```python
#find the length of the my tuple using the length function 
print(len(tuple_5))
```

```python
#checking the membership 
print((6, 8, 2) in tuple_5) #give the true if the present 
```

```python
#crating tuple 
tu_1 = (3,4,6,4)
tu_2 = (2,4,6,4)
print(tu_1+tu_2)
print(tu_2*2)
print(tu_2)
```

```python
#find the element #using the index and the count both are the use
print(tu_2.index(6))
#find the element of the repletion using the count 
print(tu_2.count(4))
```

```python
#tuple unpacking 
a,b,c,d = tu_2
print(a,b,c,d)
```

```python
#tuple swiping 
tu_3  =(40,67)
x,y = tu_3 
x,y = y,x
tu_3 = x,y
print(x,y)
print(tu_3)
```

```python
#iterating tuples 
for i in tuple_1:
    print(i)
```

```python
#converting tuple into list 
print(list(tu_1))
```

```python
#tuple using the decenary key 
dic_1 = {(4,5):"shivam" , (3,6):"itisha"}
#using the dictionary 
print(dic_1[(3,6)])
```

```python
#using the zip function in the tuple 
#creating the two tuple and the zip the element together 
tup_1 = (1,3,5,6)
tup_2 = ("mango" , "orange" , "apple " , " mago")

#now we are the zip together 
print(tuple(zip(tup_1, tup_2)))
```
