```python
#tuple using the decenary key 
dic_1 = {(4,5):"shivam" , (3,6):"itisha"}
#using the dictionary 
print(dic_1[(3,6)])
```

```python
#using the zip function in the tuple 
#creating the two tuple and the zip the element together 
tup_1 = (1,3,5,6)
tup_2 = ("mango" , "orange" , "apple " , " mago")

#now we are the zip together 
print(tuple(zip(tup_1, tup_2)))
```

```python
#creating the map function in the python using the lambda function 
tup_3 = (4,6,4,3,6)
print(tuple(map(lambda x : x**2 , tu_3)))
```

```python
tu_3 = (4,6,4,3,6,5)
#find the mix max shorted and sum of the tuple 
print(min(tu_3))
print(max(tu_3))
print(sum(tu_3))
print(tuple(sorted(tu_3)))
print(sorted(tu_3))
print(tuple(sorted(tu_3 , reverse=True)))
```

```python
#creating the nested counstoctur mix dic 
dis_2  = {1:"shiam" , 2:"itisha"  ,3:"orage"}
for i in dis_2:
    print(dis_2[i])

#creating the discretionary using the 
dis_3 = dict(name = "shivam" , id = 23 , age = 34 , Male = True)
print(dis_3)

#creating the nested 
dis_4 = {
    1 : {1:"shiam_1" , 2:"itisha"  ,3:"orage"},
    2:{1:"shia2" , 2:"itisha"  ,3:"orage"},
    3:{1:"shiam3" , 2:"itisha"  ,3:"orage"},
    4 : {1:"shiam4" , 2:"itisha"  ,3:"orage"}
}

print(dis_4)
```

```python
#2. Accessing Dictionary Values
print(dis_4[1])
print(dis_4[1][1])
#accessing using the get function 
print(dis_4.get(1))
```

```python
#updating the key values pairs 
dis_4[1][5] = "isu" ; 
print(dis_4)
```
